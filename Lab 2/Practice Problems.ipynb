## Problem 1 – Course Marks.

 Create a list containing the marks of five courses. Display the total,
average, highest mark, and lowest mark. Count the number of passed courses, using 50 as the pass mark.


```python

marks = [75, 87, 45, 91, 68]

total = sum(marks)

average = total / len(marks)

highest = max(marks)
lowest = min(marks)

passed = 0

for mark in marks:
    if mark >= 50:
        passed = passed + 1

print("Total marks:", total)

print("Average marks:", average)

print("Highest mark:", highest)

print("Lowest mark:", lowest)

print("Passed courses:", passed)
```

## Problem 3 – Word Analyzer.

Ask the user to enter a word or short sentence. Display the number
of characters, vowels (a, e, i, o, u), and spaces. Use a for loop to examine each character.


```python

text = input("Enter a word or sentence: ")

characters = len(text)

vowels = 0
spaces = 0

for char in text:
    if char.lower() in "aeiou":
        vowels = vowels + 1

    if char == " ":
        spaces = spaces + 1

print("Number of characters:", characters)

print("Number of vowels:", vowels)

print("Number of spaces:", spaces)
```

## Problem 4 – Grade Calculator.

Write a function named calculate_grade(marks) that receives a
list of marks, calculates the average, and returns the corresponding grade.

```python
def calculate_grade(marks):

    average = sum(marks) / len(marks)

    print("Average marks:", average)

    if average >= 80:
        return "A"

    elif average >= 70:
        return "B"

    elif average >= 60:
        return "C"

    elif average >= 50:
        return "D"

    else:
        return "F"


marks = [85, 75, 90, 65, 80]

grade = calculate_grade(marks)

print("Grade:", grade)
```

## Problem 5 – Shopping Cart Processor.

Create a list of dictionaries representing four products.
Each dictionary must contain a product name, price, and quantity.
Use a loop to calculate the total cost. Write a function named apply_discount(total) that applies a
10% discount when the total is at least 2,000. Otherwise, return the original total.

```python


products = [
    {"name": "Notebook", "price": 120, "quantity": 3},

    {"name": "Pen", "price": 20, "quantity": 5},

    {"name": "USB Drive", "price": 850, "quantity": 2},

    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0

for product in products:

    cost = product["price"] * product["quantity"]

    total = total + cost


def apply_discount(total):

    if total >= 2000:

        return total * 0.90

    else:
        return total


final_price = apply_discount(total)


print("Total cost:", total)

print("Final price:", final_price)
```

## Problem 6 – Countdown.

Write a program using a while loop that asks the user for a non-negative
integer and prints a countdown from that number to zero.

## Problem 7 – Student Record Analyzer.

Using the student records above, display each student’s
name and marks. Count the students who passed (marks ≥ 50), calculate the class average, and identify
the student with the highest mark.

```python
students = [

    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},

    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},

    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},

    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},

    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks = 0
passed = 0
highest_marks = 0
highest_achiever = ""

for student in students:

    print("Name:", student["name"])
    print("Marks:", student["marks"])

    total_marks = total_marks + student["marks"]

    if student["marks"] >= 50:
        passed = passed + 1

    if student["marks"] > highest_marks:

        highest_marks = student["marks"]
        highest_achiever = student["name"]

average = total_marks / len(students)

print("Number of passed students:", passed)

print("Class average:", average)

print("Student with highest marks:", highest_achiever)

print("Highest marks:", highest_marks)
```

## Problem 8 – Product Inventory Summary.

Calculate each product’s line total using price *
quantity. Calculate the total inventory value and identify the product with the highest line total.

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},

    {"name": "Pen", "price": 20, "quantity": 5},

    {"name": "USB Drive", "price": 850, "quantity": 2},

    {"name": "Mouse", "price": 650, "quantity": 1}
]

total_inventory = 0
highest_total = 0
highest_product = ""

for product in products:

    line_total = product["price"] * product["quantity"]

    print("Product:", product["name"])

    print("Line total:", line_total)

    total_inventory = total_inventory + line_total

    if line_total > highest_total:

        highest_total = line_total

        highest_product = product["name"]

print("Total inventory value:", total_inventory)

print("Product with highest line total:", highest_product)

print("Highest line total:", highest_total)
```
