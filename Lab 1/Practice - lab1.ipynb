## Problem 1.

Write a program that inputs two numbers and displays their sum, difference, product, and
quotient. Convert the inputs to numeric values before calculating.

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

sum_num = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2

print("Sum:", sum_num)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
```

Input a student’s name, student ID, age, department, and CGPA. Display the information

## Problem 3.

Input a temperature in Celsius and convert it to Fahrenheit and Kelvin using the formulas:
F = (C × 9/5) + 32, K = C + 273.15

```python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("Temperature in Celsius:", celsius)
print("Temperature in Fahrenheit:", fahrenheit)
print("Temperature in Kelvin:", kelvin)
```

## Problem 4.

Input a student’s name and five course marks. Calculate the total, average, highest mark,
lowest mark, and number of passed courses. Display all results. A mark of 50 or higher is considered a
pass.

```python
name = input("Enter student's name: ")

marks = []

for i in range(5):
    mark = int(input("Enter mark for course: "))
    marks.append(mark)

total = 0
highest = marks[0]
lowest = marks[0]
passed = 0

for mark in marks:
    total += mark

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

    if mark >= 50:
        passed += 1

average = total / 5

print("\nStudent Name:", name)
print("Total Marks:", total)
print("Average Mark:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Number of Passed Courses:", passed)
```
